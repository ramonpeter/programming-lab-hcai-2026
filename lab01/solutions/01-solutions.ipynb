# Part B — Three ways to run Python

### B1 · 🟢 core · The interactive prompt (REPL)

In a terminal type `python` (Windows) or `python3` (macOS/Linux). You are now *inside*
Python — the prompt changes to `>>>`.

1. Type `2 ** 10` and press Enter.
2. Type `17 / 4`.
3. Type `python --version`. **What happens, and why?**
4. Leave with `exit()`.

**Solution.** 3 gives `NameError: name 'python' is not defined`. `python --version` is a command for the
**terminal**, not a line of Python — so Python reads it as "the variable `python`, minus minus
`version`" and has never heard of either. The prompt tells you who you are talking to: `>>>` is Python;
`$`, `%` or `C:\...>` is the terminal. Mixing the two up is the most common first-week confusion.

### B2 · 🟢 core · A script

In VS Code, create a new file `hello.py` in `lab01` with exactly these two lines:

```python
print("Hello from my first script")
2 ** 10
```

Run it twice: once with the ▷ button (top right), once from the terminal with
`python hello.py` (or `python3 hello.py`).

**What is printed? Where is the 1024?**

Now paste the same two lines into the empty cell below and run it. **What is different?**

````python
print("Hello from my first script")
2 ** 10
````

**Solution.** The script prints only the greeting. `2 ** 10` is computed and then thrown away — a script
shows only what you `print`. A notebook cell (and the REPL) additionally *displays the value
of the last line*, which is why `1024` appears here. This is a convenience of notebooks, not
something Python does. Programs you hand in, or run on a server, are scripts: **if you want
to see it, print it.**

### C4 · 🔴 challenge · Big numbers

Run `print(2 ** 100)` and `print(2.0 ** 100)`. Explain the difference in the two outputs.
Then: how many seconds are there in a (non-leap) year? And roughly how many years is
`2 ** 64` seconds?

````python
print(2 ** 100)
print(2.0 ** 100)
````

````python
print("Seconds per year:", 365 * 24 * 60 * 60)
print("2**64 seconds in years:", 2 ** 64 / (365 * 24 * 60 * 60))
````

**Solution.** Python whole numbers (`int`) have no size limit, so `2 ** 100` is exact. A decimal number

### D4 · 🔴 challenge · A mistake that does not crash

Someone wanted 2 to the power of 10 and wrote:

````python
print(2 ^ 10)
````

It runs. It prints `8`. Find out what `^` does in Python (search
<https://docs.python.org/3/reference/expressions.html> for `^`) and explain why the answer
is 8.

**Solution.** `^` is **bitwise exclusive or**, not power. In binary, 2 is `0010` and 10 is `1010`; XOR keeps
the bits that differ: `1000`, which is 8. The lesson is bigger than the operator: **a program
that runs and gives a wrong number is more dangerous than one that crashes.** Always ask
whether an answer is plausible.
