# Part A — Names and values

A variable is a **name** attached to a **value**. `=` does not mean "equals": it means
*compute the right-hand side, then attach the name on the left to the result*.

### A1 · 🟢 core · Trace it by hand

Without running anything, fill in the table: the value of each name **after** each line.
Then run the cell and check.

| after line | `a` | `b` | `c` |
|---|---|---|---|
| 1 | | | |
| 2 | | | |
| 3 | | | |
| 4 | | | |
| 5 | | | |

````python
a = 4           # line 1
b = a * 2       # line 2
a = 10          # line 3
c = a + b       # line 4
b = b - 1       # line 5
print(a, b, c)
````

**Solution.** 
| after line | `a` | `b` | `c` |
|---|---|---|---|
| 1 | 4 | – | – |
| 2 | 4 | 8 | – |
| 3 | 10 | 8 | – |
| 4 | 10 | 8 | 18 |
| 5 | 10 | 7 | 18 |

Printed: `10 7 18`. The point is line 3: changing `a` does **not** change `b`. Line 2 computed
`a * 2` once, got `8`, and attached `b` to that value. `b` has no memory of where 8 came from.

# Part B — Types and conversion

Every value has a **type**: `int` (whole number), `float` (decimal number), `str` (text),
`bool` (`True`/`False`). `type(x)` tells you which.

### B1 · 🟢 core · Predict the type

Write the type you expect as a comment, then run.

````python
print(type(7))          # int
print(type(7.0))        # float   — the .0 makes it a float
print(type("7"))        # str     — quotes make text, whatever is inside
print(type(True))       # bool
print(type(7 / 7))      # float   — / always gives a float (1.0)
print(type(7 // 7))     # int     — // of two ints is an int (1)
print(type(7 * 1.0))    # float   — int with float gives float
print(type("7" * 2))    # str     — "77"
print(type(3 > 2))      # bool    — a comparison is a question with a yes/no answer
print(type("False"))    # str     — in quotes, it is just five letters
````

### B5 · 🔴 challenge · Rounding surprise

up `round` in the documentation (<https://docs.python.org/3/library/functions.html#round>).

# Part C — Expressions

Order: brackets → `**` → `*` `/` `//` `%` → `+` `-`; equal levels go left to right
(except `**`). `x += 3` is short for `x = x + 3` — same for `-=`, `*=`, `/=`, `//=`.

### C1 · 🟢 core · Predict, then run

````python
print(2 + 3 * 4 ** 2)        # 50   4**2 = 16 first, then 3*16, then +2
print((2 + 3) * 4 ** 2)      # 80
print(10 - 4 - 3)            # 3    left to right: (10-4)-3, not 10-(4-3)
print(-(2 ** 2))               # -4   ** binds tighter than the minus sign: -(2**2)
print((-2) ** 2)             # 4
print(2 ** -1)               # 0.5
print(17 // 5 * 5 + 17 % 5)  # 17   3*5 + 2: // and % together give the number back
````

### C3 · 🟢 core · Pizza for the lab

Write it with **variables**, so the numbers can change:

`people = 23`, each person eats `3` slices, a pizza has `8` slices.
Print how many slices are needed, how many **whole pizzas** must be ordered, and how many
slices are left over.

*Hint:* `slices // 8` rounds down. You need to round **up**. Try `(slices + 7) // 8`, and
work out why it does that.

````python
people = 23
slices_each = 3
slices_per_pizza = 8

slices = people * slices_each
pizzas = (slices + slices_per_pizza - 1) // slices_per_pizza   # round up
leftover = pizzas * slices_per_pizza - slices
leftover2 = (slices + slices_per_pizza - 1) % slices_per_pizza   # alternative: use the remainder operator

print("Slices needed:", slices)
print("Pizzas to order:", pizzas)
print("Slices left over:", leftover)
print("Slices left over:", leftover2)
````

69 slices → 9 pizzas → 3 left over. Adding `8 - 1` before `// 8` pushes every non-exact case over
the next multiple of 8, but leaves exact multiples alone (64 + 7 = 71, still 8 pizzas).
Test it: change `people` to 8 (24 slices: exactly 3 pizzas, 0 left over).

### D2 · 🟢 core · Split the bill

Create `bill.py`. Ask for: the total bill in euro, the number of people, and the tip in percent.
Print the amount per person, rounded to 2 decimals.

```
Total bill (EUR): 86.40
Number of people: 4
Tip in percent: 10
Each person pays: 23.76 EUR
```

````python
total = float(input("Total bill (EUR): "))
people = int(input("Number of people: "))
tip_percent = float(input("Tip in percent: "))

if people < 0:
    print("Error: number of people must be positive.")
    
else:
    with_tip = total * (1 + tip_percent / 100)
    per_person = with_tip / people
    print("Each person pays:", round(per_person, 2), "EUR")
````

Why `float` for the bill but `int` for people: half a person does not split a bill.
`round(..., 2)` drops a trailing zero (`23.8` instead of `23.80`); f-strings fix that — see E2.

### D5 · 🔴 challenge · How many photos?

A phone has 128 GB, of which the system uses 14 GB. One photo is 4.2 MB. Ask for the number of
photos already on the phone and print how many more fit. Use 1 GB = 1000 MB. Is your answer a
whole number? Should it be?

````python
TOTAL_GB = 128
SYSTEM_GB = 14
PHOTO_MB = 4.2

stored = int(input("Photos already on the phone: "))
free_mb = (TOTAL_GB - SYSTEM_GB) * 1000 - stored * PHOTO_MB
more = int(free_mb // PHOTO_MB)
print("Room for", more, "more photos")
````

With 0 stored: `114000 // 4.2` is `27142.0` — `//` with a float gives a float, so `int(...)` makes
it a whole number of photos. With 30000 stored the result is negative: the input is
impossible, and the program cannot tell yet. Another one for session 3.

### E3 · 🔴 challenge · Your real week

Not everyone studies 7 days. Ask how many days the user wants to study, and also how many hours
of sleep they need per night. Print the hours per study day, and what fraction of their waking
time that is, as a percentage.

### Solution: `assistant.py` v0.2

````python
# Name: Ada Lovelace
# Date: 12.10.2026
# A rule-based study assistant. v0.2 asks for this week's numbers.

SELF_STUDY_FACTOR = 1.5   # hours of self-study per lecture hour
DAYS = 7

print("=" * 50)
print("   STUDY ASSISTANT  v0.2")
print("=" * 50)
name = input("What is your name? ")
lectures = int(input("How many lectures this week? "))
hours_per_lecture = float(input("How many hours is one lecture? "))

lecture_hours = lectures * hours_per_lecture
self_study_hours = lecture_hours * SELF_STUDY_FACTOR
total_hours = lecture_hours + self_study_hours
hours_per_day = total_hours / DAYS

print()
print("Hello,", name + "! Your week:")
print("Lecture hours:", lecture_hours)
print("Self-study hours (" + str(SELF_STUDY_FACTOR) + " per lecture hour):", self_study_hours)
print("Total hours:", total_hours)
print("Hours per day (" + str(DAYS) + " days):", round(hours_per_day, 1))
print("=" * 50)

# 7 lectures next week: no line changes — you type 7 when it asks.
````

### Solution: E2 with f-strings (the report part)

````python
print(f"Hello, {name}! Your week:")
print(f"Lecture hours: {lecture_hours:.1f}")
print(f"Self-study hours ({SELF_STUDY_FACTOR} per lecture hour): {self_study_hours:.1f}")
print(f"Total hours: {total_hours:.1f}")
print(f"Hours per day ({DAYS} days): {hours_per_day:.1f}")
````

### Solution: E3 (one way)

````python
study_days = int(input("How many days do you want to study? "))
sleep = float(input("Hours of sleep per night? "))

per_study_day = total_hours / study_days
waking = 24 - sleep
print(f"Hours per study day: {per_study_day:.1f}")
print(f"That is {per_study_day / waking * 100:.0f}% of your waking time.")
````
