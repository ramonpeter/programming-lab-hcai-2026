# Part C — Expressions

Order: brackets → `**` → `*` `/` `//` `%` → `+` `-`; equal levels go left to right
(except `**`). `x += 3` is short for `x = x + 3` — same for `-=`, `*=`, `/=`, `//=`.

### C1 · 🟢 core · Predict, then run

````python
print(2 + 3 * 4 ** 2)        # prediction:
print((2 + 3) * 4 ** 2)      # prediction:
print(10 - 4 - 3)            # prediction:
print(-2 ** 2)               # prediction:
print((-2) ** 2)             # prediction:
print(2 ** -1)               # prediction:
print(17 // 5 * 5 + 17 % 5)  # prediction:
````
