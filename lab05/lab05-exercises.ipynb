**0**, and negative indices that count from the end (`-1` is the last character).
`s[start:stop:step]` takes a slice: `start` is included, `stop` is **not**.

# Part B — String methods

A method is a function that belongs to a value: `text.lower()`. String methods never change
the string (they cannot — strings are immutable). They **return a new one**, so you have to keep
the result: `text = text.lower()`.

### B1 · 🟢 core · Predict, then run

````python
t = "  Hello, Lab!  "
print(repr(t.strip()))                  # prediction:
print(t.lower())                        # prediction:
print(t.strip().upper())                # prediction:
print(t.replace("l", "L"))              # prediction:
print(t.find("Lab"))                    # prediction:
print(t.find("lab"))                    # prediction:
print(t.count("l"))                     # prediction:
print(t.strip().startswith("Hello"))    # prediction:
print(t.split(","))                     # prediction:
print("2026".isdigit(), "20.5".isdigit())   # prediction:
````

# Part D — Tuples and sets

| | list `[ ]` | tuple `( )` | set `{ }` |
|---|---|---|---|
| ordered, indices | yes | yes | no |
| can change | yes | **no** | yes (add, remove) |
| duplicates | allowed | allowed | **never** |
| use it for | a collection that grows | a fixed record: a point, a pair | membership, uniqueness, comparing groups |

# Part E — `todo.py`: a tiny command interpreter

A program that understands four commands — like a chatbot with a very small vocabulary, and no
learning at all. Write it as a script `todo.py` in `lab05` and run it from the terminal:

```
> add read chapter 3
Added: read chapter 3
> add exercise sheet
Added: exercise sheet
> list
1. read chapter 3
2. exercise sheet
> done read chapter 3
Done: read chapter 3
> done laundry
Not on the list: laundry
> quit
Bye — 1 task left.
```

### E1 · 🟢 core · Four commands

- Keep the tasks in a **list**. Loop with `while True:` and leave with `break` on `quit`.
- Read a line with `input("> ")`, clean it with `.strip()`, and split it **once** into command and
  rest: `parts = line.split(" ", 1)`. The command is `parts[0].lower()`; the text, if there is one,
  is `parts[1]`.
- `add TEXT` appends, `done TEXT` removes (check with `in` first), `list` prints the tasks numbered
  from 1, `quit` stops.

### E2 · 🟡 stretch · Be forgiving

- An empty line does nothing; an unknown command prints a short help.
- `add` with a task that is already on the list (ignoring upper/lower case) says so instead of adding
  it twice.
